```python
#2.import numpy as np
import pandas as pd

# Reproducible random data
np.random.seed(42)

# Number of records
n = 5000

# Create product and category mapping
product_category = {
    "Laptop": "Electronics",
    "Mobile": "Electronics",
    "Tablet": "Electronics",
    "Headphones": "Accessories",
    "Keyboard": "Accessories",
    "Mouse": "Accessories",
    "Monitor": "Electronics",
    "Smartwatch": "Wearables"
}

# Generate products
products = np.random.choice(
    list(product_category.keys()),
    size=n
)

# Generate DataFrame
df = pd.DataFrame({
    "Order_ID": np.arange(10001, 10001 + n),

    "Customer_ID": np.random.randint(
        1001, 2001, n
    ),

    "Product": products,

    "Category": [
        product_category[p]
        for p in products
    ],

    "City": np.random.choice([
        "Delhi",
        "Mumbai",
        "Bangalore",
        "Chennai",
        "Kolkata",
        "Hyderabad",
        "Pune",
        "Patna"
    ], n),

    "Quantity": np.random.randint(
        1, 6, n
    ),

    "Price": np.random.randint(
        500, 100000, n
    ),

    "Discount": np.random.choice(
        [0, 5, 10, 15, 20],
        n
    ),

    "Payment_Method": np.random.choice([
        "UPI",
        "Credit Card",
        "Debit Card",
        "Cash",
        "Net Banking"
    ], n),

    "Order_Status": np.random.choice([
        "Delivered",
        "Delivered",
        "Delivered",
        "Delivered",
        "Cancelled",
        "Returned"
    ], n),

    "Rating": np.random.randint(
        1, 6, n
    ),

    "Order_Date": pd.to_datetime(
        np.random.choice(
            pd.date_range(
                "2025-01-01",
                "2025-12-31"
            ),
            size=n
        )
    )
})

# Display first 5 rows
df.head()
```

```python
#6. Create Sales Column
df["Gross_Sales"] = (
    df["Quantity"] * df["Price"]
)

df.head()
```

```python
#7. NumPy Analysis
total_sales = np.sum(df["Net_Sales"])

print("Total Sales:", total_sales)
```

```python
#Average Sales
average_sales = np.mean(df["Net_Sales"])

print("Average Sales:", average_sales)
```

```python
#Highest Sale
highest_sale = np.max(df["Net_Sales"])

print("Highest Sale:", highest_sale)
```

```python
#Lowest Sale
lowest_sale = np.min(df["Net_Sales"])

print("Lowest Sale:", lowest_sale)
```

```python
#Median
median_sales = np.median(df["Net_Sales"])

print("Median Sales:", median_sales)
```

```python
#Standard Deviation
std_sales = np.std(df["Net_Sales"])

print("Standard Deviation:", std_sales)
```

```python
#8. Highest Sales Order
index = np.argmax(df["Net_Sales"])

df.iloc[index]
```

```python
#Lowest:
index = np.argmin(df["Net_Sales"])

df.iloc[index]
```

```python
#9. Product Analysis
product_sales = (
    df.groupby("Product")["Net_Sales"]
    .sum()
    .sort_values(ascending=False)
)

product_sales
```

```python
#Top 5 Products
product_sales.head(5)
```

```python
#Product Quantity
product_quantity = (
    df.groupby("Product")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

product_quantity
```

```python
#10. Category Analysis
category_analysis = (
    df.groupby("Category")
    .agg(
        Total_Sales=("Net_Sales", "sum"),
        Average_Sales=("Net_Sales", "mean"),
        Total_Quantity=("Quantity", "sum"),
        Total_Orders=("Order_ID", "count")
    )
    .sort_values(
        "Total_Sales",
        ascending=False
    )
)

category_analysis
```

```python
#11. City Analysis
city_sales = (
    df.groupby("City")["Net_Sales"]
    .sum()
    .sort_values(ascending=False)
)

city_sales
```

```python
#Top city:
top_city = city_sales.idxmax()

print("Top City:", top_city)
```

```python
#12. Customer Analysis
customer_sales = (
    df.groupby("Customer_ID")["Net_Sales"]
    .sum()
    .sort_values(ascending=False)
)

customer_sales.head(10)
```

```python
#Customer Order Count
customer_orders = (
    df.groupby("Customer_ID")["Order_ID"]
    .count()
    .sort_values(ascending=False)
)

customer_orders.head(10)
```

```python
#13. Payment Method Analysis
payment_analysis = (
    df.groupby("Payment_Method")
    .agg(
        Total_Sales=("Net_Sales", "sum"),
        Orders=("Order_ID", "count"),
        Average_Order=("Net_Sales", "mean")
    )
    .sort_values(
        "Total_Sales",
        ascending=False
    )
)

payment_analysis
```

```python
#14. Order Status Analysis
status_analysis = (
    df.groupby("Order_Status")["Order_ID"]
    .count()
)

status_analysis
```

```python
#Cancellation rate
cancelled = np.sum(
    df["Order_Status"] == "Cancelled"
)

total_orders = len(df)

cancellation_rate = (
    cancelled / total_orders
) * 100

print(
    "Cancellation Rate:",
    round(cancellation_rate, 2),
    "%"
)
```

```python
#Return rate:
returned = np.sum(
    df["Order_Status"] == "Returned"
)

return_rate = (
    returned / total_orders
) * 100

print(
    "Return Rate:",
    round(return_rate, 2),
    "%"
)
```

```python
#15. Rating Analysis
rating_analysis = (
    df.groupby("Rating")["Order_ID"]
    .count()
)

rating_analysis
```

```python
#Average rating:
average_rating = df["Rating"].mean()

print(
    "Average Rating:",
    round(average_rating, 2)
)
```

```python
#16. Sales Above Average
average_sales = df["Net_Sales"].mean()

high_value_orders = df[
    df["Net_Sales"] > average_sales
]

high_value_orders.head()
```

```python
#17. NumPy where()
df["Sales_Level"] = np.where(
    df["Net_Sales"] >= average_sales,
    "High",
    "Low"
)

df[
    ["Order_ID", "Net_Sales", "Sales_Level"]
].head()
```

```python
#18. Outlier Analysis
Q1 = np.percentile(
    df["Net_Sales"], 25
)

Q3 = np.percentile(
    df["Net_Sales"], 75
)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
#Find outliers:
outliers = df[
    (df["Net_Sales"] < lower_limit) |
    (df["Net_Sales"] > upper_limit)
]

print("Number of Outliers:", len(outliers))
```
