```python
#1. Basic while loop
i = 0

while i < len(df):
    print(df.loc[i, "Employee"])
    i += 1
```

```python
#2. Employee Sales Report
i = 0

while i < len(df):
    name = df.loc[i, "Employee"]
    sales = df.loc[i, "Sales"]

    print(name, ":", sales)

    i += 1
```

```python
#3. Find High Sales
i = 0

while i < len(df):
    name = df.loc[i, "Employee"]
    sales = df.loc[i, "Sales"]

    if sales > 50000:
        print(name, "-", sales)

    i += 1
```

```python
#4. Calculate Total Sales
i = 0
total_sales = 0

while i < len(df):
    total_sales += df.loc[i, "Sales"]
    i += 1

print("Total Sales:", total_sales)
```

```python
#5. Calculate Total Orders
i = 0
total_orders = 0

while i < len(df):
    total_orders += df.loc[i, "Orders"]
    i += 1

print("Total Orders:", total_orders)
```

```python
#6. Calculate Average Sales
i = 0
total_sales = 0

while i < len(df):
    total_sales += df.loc[i, "Sales"]
    i += 1

average_sales = total_sales / len(df)

print("Average Sales:", average_sales)
```

```python
#7. Find Highest Sales
i = 0
highest_sales = df.loc[0, "Sales"]
highest_employee = df.loc[0, "Employee"]

while i < len(df):
    if df.loc[i, "Sales"] > highest_sales:
        highest_sales = df.loc[i, "Sales"]
        highest_employee = df.loc[i, "Employee"]

    i += 1

print("Highest Sales:", highest_sales)
print("Employee:", highest_employee)
```

```python
#8. Complete while Loop Project
import pandas as pd

data = {
    "Employee": ["Rahul", "Amit", "Ravi", "Priya", "Neha", "Suresh"],
    "City": ["Patna", "Delhi", "Mumbai", "Patna", "Delhi", "Mumbai"],
    "Sales": [45000, 62000, 38000, 75000, 55000, 52000],
    "Orders": [25, 32, 20, 40, 28, 30]
}

df = pd.DataFrame(data)

i = 0
total_sales = 0
total_orders = 0
highest_sales = 0
highest_employee = ""

while i < len(df):
    name = df.loc[i, "Employee"]
    city = df.loc[i, "City"]
    sales = df.loc[i, "Sales"]
    orders = df.loc[i, "Orders"]

    total_sales += sales
    total_orders += orders

    if sales > highest_sales:
        highest_sales = sales
        highest_employee = name

    print(
        name,
        "|", city,
        "| Sales:", sales,
        "| Orders:", orders
    )

    i += 1

average_sales = total_sales / len(df)

print("\n--- KPI REPORT ---")
print("Total Employees:", len(df))
print("Total Sales:", total_sales)
print("Total Orders:", total_orders)
print("Average Sales:", round(average_sales, 2))
print("Highest Sales:", highest_sales)
print("Top Employee:", highest_employee)
```
