```python
# string function 
```

```python
#1. upper text ko capital karta hai
name = "rahul"
print(name.upper())
```

```python
#2. lower text ko small letters mein karta hai
name = "RAHUL"
print(name.lower())
```

```python
#3. title har word ka first letter capital karta hai
name = "rahul kumar"
print(name.title())
```

```python
#4. capitalize first letter capital karta hai
name = "rahul kumar"
print(name.capitalize())
```

```python
#5. strip extra spaces remove karta hai
name = "  Rahul  "
print(name.strip())
```

```python
#6. replace text ko replace karta hai
city = "Patna Bihar"
print(city.replace("Patna", "Delhi"))
```

```python
#7. split string ko parts mein todta hai
name = "Rahul Kumar"
print(name.split())
```

```python
#8. join strings ko jodta hai
names = ["Rahul", "Amit", "Ravi"]

print(", ".join(names))
```

```python
#9. find text ki position dhundta hai
email = "rahul@gmail.com"

print(email.find("@"))
```

```python
#10. count kisi text ki counting karta hai
text = "apple banana apple mango apple"

print(text.count("apple"))
```

```python
#11. startswith check karta hai ki text kis se start ho raha hai
email = "rahul@gmail.com"

print(email.startswith("rahul"))
```

```python
#12. endswith check karta hai ki text kis par end ho raha hai
email = "rahul@gmail.com"

print(email.endswith(".com"))
```

```python
#13. isdigit check karta hai ki value number hai
value = "12345"

print(value.isdigit())
```

```python
#14. isalpha check karta hai ki sirf letters hain
name = "Rahul"

print(name.isalpha())
```

```python
#15. len string ki length nikalta hai
name = "Rahul Kumar"

print(len(name))
```

```python
#16. swapcasecapital ko small aur small ko capital karta hai
text = "RaHuL KuMaR"

print(text.swapcase())
```

```python
#17. casefold text ko lowercase comparison ke liye convert karta hai
text = "HELLO WORLD"

print(text.casefold())
```

```python
#18. lstrip left side ke spaces hataata hai
text = "   Rahul"

print(text.lstrip())
```

```python
#19. rstrip() — right side ke spaces hataata hai
text = "Rahul   "

print(text.rstrip())
```

```python
#20. center() — text ko center mein rakhta hai
text = "Sales"

print(text.center(20, "-"))
```

```python
#21. zfill number ke aage zero lagata hai
order_id = "125"

print(order_id.zfill(6))
```

```python
#22. index() — text ki position batata hai
email = "rahul@gmail.com"

print(email.index("@"))
```

```python
#23. rfind() — last occurrence ki position deta hai
text = "apple banana apple"

print(text.rfind("apple"))

```

```python
#24. partition string ko 3 parts mein divide karta hai
email = "rahul@gmail.com"

print(email.partition("@"))
```

```python
#25. rpartition right side se partition karta hai
text = "sales-report-2026"

print(text.rpartition("-"))
```
