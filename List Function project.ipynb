```python
#list function
```

```python
#1. append Add one item
sales = [100, 200, 300]

sales.append(400)

print(sales)
```

```python
#2. insert Add an item at a position
sales = [100, 200, 300]

sales.insert(1, 150)

print(sales)
```

```python
#3. extend Add multiple items
sales = [100, 200, 300]

sales.extend([400, 500])

print(sales)
```

```python
#4. remove Remove a specific value
sales = [100, 200, 300, 400]

sales.remove(300)

print(sales)
```

```python
#5. pop Remove an item by index
sales = [100, 200, 300, 400]

sales.pop(1)

print(sales)
```

```python
#6. sort Sort data
sales = [500, 100, 400, 200, 300]

sales.sort()

print(sales)
```

```python
#7. reverse Reverse the list
sales = [100, 200, 300, 400]

sales.reverse()

print(sales)
```

```python
#8. count Count a value
sales = [100, 200, 100, 300, 100]

print(sales.count(100))
```

```python
#9. index Find the position
sales = [100, 200, 300, 400]

print(sales.index(300))
```

```python
#10. len Count total items
sales = [100, 200, 300, 400]

print(len(sales))
```

```python
#11. sum Calculate tota
sales = [100, 200, 300, 400]

print(sum(sales))
```

```python
#12. max Find the highest value
sales = [500, 200, 800, 100, 400]

print(max(sales))
```

```python
#13. min Find the lowest value
sales = [500, 200, 800, 100, 400]

print(min(sales))
```

```python
#14. Slicing  Get a part of the list
sales = [100, 200, 300, 400, 500]

print(sales[1:4])
print(sales[:3])
print(sales[2:])
```

```python
#15. List Comprehension  Filter data
sales = [100, 250, 300, 450, 500]

high_sales = [x for x in sales if x > 300]

print(high_sales)
```
