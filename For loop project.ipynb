```python
#1. Print every employee
for employee in df["Employee"]:
    print(employee)
```

```python
#2. Print employee and city
for i in range(len(df)):
    print(df.loc[i, "Employee"], "-", df.loc[i, "City"])
```

```python
#3. Print employee and sales
for i in range(len(df)):
    print(df.loc[i, "Employee"], "-", df.loc[i, "Sales"])
```

```python
#4. Find employees with sales above 50,000
for i in range(len(df)):
    if df.loc[i, "Sales"] > 50000:
        print(df.loc[i, "Employee"])
```

```python
#5. Calculate sales per order
for i in range(len(df)):
    sales = df.loc[i, "Sales"]
    orders = df.loc[i, "Orders"]
    print(df.loc[i, "Employee"], round(sales / orders, 2))
```

```python
#6. Print complete employee report
for i in range(len(df)):
    print(
        i + 1,
        df.loc[i, "Employee"],
        df.loc[i, "City"],
        df.loc[i, "Sales"],
        df.loc[i, "Orders"]
    )
```

```python
#7. Find high-order employees
for i in range(len(df)):
    if df.loc[i, "Orders"] > 30:
        print(df.loc[i, "Employee"], df.loc[i, "Orders"])
```

```python
#8. Create a simple KPI report using for
total_sales = 0

for sales in df["Sales"]:
    total_sales += sales

print("Total Sales:", total_sales)
```

```python
#9. City-wise employee report
for city in df["City"].unique():
    count = 0

    for value in df["City"]:
        if value == city:
            count += 1

    print(city, ":", count, "employees")
```

```python
#10. Final project-style loop
print("EMPLOYEE SALES REPORT")
print("-" * 40)

for i in range(len(df)):
    name = df.loc[i, "Employee"]
    city = df.loc[i, "City"]
    sales = df.loc[i, "Sales"]
    orders = df.loc[i, "Orders"]

    if sales > 60000:
        status = "High Performer"
    elif sales > 45000:
        status = "Average Performer"
    else:
        status = "Low Performer"

    print(name, "|", city, "| Sales:", sales, "| Orders:", orders, "|", status)
```
