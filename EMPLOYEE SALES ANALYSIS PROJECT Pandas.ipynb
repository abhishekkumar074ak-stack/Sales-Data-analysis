```python
#EMPLOYEE SALES ANALYSIS PROJECT
```

```python
# 1. Dictionary
employees = {
    "Rahul": {
        "city": "Patna",
        "sales": 45000,
        "orders": 25
    },
    "Amit": {
        "city": "Delhi",
        "sales": 62000,
        "orders": 32
    },
    "Ravi": {
        "city": "Mumbai",
        "sales": 38000,
        "orders": 20
    },
    "Priya": {
        "city": "Patna",
        "sales": 75000,
        "orders": 40
    },
    "Neha": {
        "city": "Delhi",
        "sales": 55000,
        "orders": 28
    }
}
```

```python
# 2. keys()
print("Employee Names:")
print(employees.keys())
```

```python
# 3. values()
print("\nEmployee Data:")
print(employees.values())
```

```python
# 4. items()
print("\nEmployee Details:")

for name, data in employees.items():
    print(name, data)
```

```python
# 5. Nested Dictionary
print("\nRahul's Data:")
print(employees["Rahul"])
```

```python
# 6. Total Sales using sum()
total_sales = sum(
    data["sales"]
    for data in employees.values()
)

print("\nTotal Sales:", total_sales)
```

```python
# 7. Average Sales
average_sales = total_sales / len(employees)

print("Average Sales:", average_sales)
```

```python
# 8. Maximum Sales
highest_sales = max(
    data["sales"]
    for data in employees.values()
)

print("Highest Sales:", highest_sales)
```

```python
# 9. Minimum Sales
lowest_sales = min(
    data["sales"]
    for data in employees.values()
)

print("Lowest Sales:", lowest_sales)
```

```python
# 10. Lambda
highest_employee = max(
    employees,
    key=lambda name: employees[name]["sales"]
)

print("Highest Sales Employee:", highest_employee)
```

```python
# 11. Minimum using Lambda
lowest_employee = min(
    employees,
    key=lambda name: employees[name]["sales"]
)

print("Lowest Sales Employee:", lowest_employee)
```

```python
# 12. For Loop
print("\nAll Employees:")

for name, data in employees.items():
    print(
        name,
        "=>",
        data["city"],
        "| Sales:", data["sales"],
        "| Orders:", data["orders"]
    )
```

```python
# 13. If Condition
print("\nEmployees with Sales Above 50,000:")

for name, data in employees.items():

    if data["sales"] > 50000:
        print(name, data["sales"])
```

```python
# 14. Data Filtering
high_sales_employees = {
    name: data
    for name, data in employees.items()
    if data["sales"] > 50000
}

print("\nFiltered Employees:")
print(high_sales_employees)
```

```python
# 15. Filter Employees from Patna
patna_employees = {
    name: data
    for name, data in employees.items()
    if data["city"] == "Patna"
}

print("\nEmployees from Patna:")
print(patna_employees)
```

```python
# 16. Total Orders
total_orders = sum(
    data["orders"]
    for data in employees.values()
)

print("\nTotal Orders:", total_orders)
```

```python
# 17. Average Orders
average_orders = total_orders / len(employees)

print("Average Orders:", average_orders)
```

```python
# 18. Sales Per Order
print("\nSales Per Order:")

for name, data in employees.items():

    sales_per_order = data["sales"] / data["orders"]

    print(
        name,
        ":",
        round(sales_per_order, 2)
    )
```

```python

# 19. Data Updating
employees["Rahul"]["sales"] = 50000

print("\nUpdated Rahul Data:")
print(employees["Rahul"])
```

```python

# 20. Add New Employee
employees["Suresh"] = {
    "city": "Kolkata",
    "sales": 48000,
    "orders": 24
}

print("\nNew Employee:")
print(employees["Suresh"])
```
