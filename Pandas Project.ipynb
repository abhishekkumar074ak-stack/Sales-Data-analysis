```python
#2. Create Data
data = {
    "Employee": ["Rahul", "Amit", "Ravi", "Priya", "Neha", "Suresh"],
    "City": ["Patna", "Delhi", "Mumbai", "Patna", "Delhi", "Mumbai"],
    "Sales": [45000, 62000, 38000, 75000, 55000, 52000],
    "Orders": [25, 32, 20, 40, 28, 30]
}
```

```python
#3. Create DataFrame
df = pd.DataFrame(data)
print(df)
```

```python
#4. Check Data Type
print(type(df))
```

```python
#5. Count Employees
print("Total Employees:", len(df))
```

```python
#6. Check First 5 Rows
print(df.head())
```

```python
#7. Check Information
df.info()
```

```python
#8. Statistical Summary
print(df.describe())
```

```python
#9. Check Shape
print("Shape:", df.shape)
```

```python
#10. Calculate Total Sales
total_sales = df["Sales"].sum()
print("Total Sales:", total_sales)
```

```python
#11. Find Highest Sales
highest_sales = df["Sales"].max()
print("Highest Sales:", highest_sales)
```

```python
#12. Find Lowest Sales
lowest_sales = df["Sales"].min()
print("Lowest Sales:", lowest_sales)
```

```python
#13. Calculate Average Sales
average_sales = df["Sales"].mean()
print("Average Sales:", average_sales)
```

```python
#14. Filter High Sales
high_sales = df[df["Sales"] > 50000]
print(high_sales)
```

```python
#15. Group Sales by City
city_sales = df.groupby("City")["Sales"].sum()
print(city_sales)
```

```python
#16. Group Orders by City
city_orders = df.groupby("City")["Orders"].sum()
print(city_orders)
```

```python
#17. Final KPI Report
print("Total Employees:", len(df))
print("Total Sales:", df["Sales"].sum())
print("Total Orders:", df["Orders"].sum())
print("Average Sales:", round(df["Sales"].mean(), 2))
print("Highest Sales:", df["Sales"].max())
print("Lowest Sales:", df["Sales"].min())
```
