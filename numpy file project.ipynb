```python
#1.Import NumPy
import numpy as np
```

```python
#2.Create Sales Dataset
sales = np.array([
    12000, 15000, 18000, 11000, 22000,
    25000, 17000, 30000, 28000, 19000
])

print(sales)
```

```python
#3.Basic Information
print("Number of records:", sales.size)
print("Shape:", sales.shape)
print("Data type:", sales.dtype)
```

```python
#4. Total, Average, Highest, Lowest Sales
print("Total Sales:", np.sum(sales))
print("Average Sales:", np.mean(sales))
print("Highest Sales:", np.max(sales))
print("Lowest Sales:", np.min(sales))
```

```python
#5. Median and Standard Deviation
print("Median Sales:", np.median(sales))
print("Standard Deviation:", np.std(sales))
print("Variance:", np.var(sales))
```

```python
#6. Find Highest and Lowest Sale Position
highest_index = np.argmax(sales)
lowest_index = np.argmin(sales)

print("Highest Sale:", sales[highest_index])
print("Highest Sale Index:", highest_index)

print("Lowest Sale:", sales[lowest_index])
print("Lowest Sale Index:", lowest_index)
```

```python
#7. Filtering Data
high_sales = sales[sales > 20000]

print(high_sales)
```

```python
medium_sales = sales[(sales >= 15000) & (sales <= 25000)]

print(medium_sales)
```

```python
#8. Count Sales Above ₹20,000
count = np.sum(sales > 20000)

print("Sales above 20000:", count)
```

```python
#9. Percentage Calculation
discount = sales * 0.10

final_sales = sales - discount

print(final_sales)
```

```python
#10. Sales Growth
previous_sales = np.array([
    10000, 14000, 16000, 12000, 20000,
    23000, 15000, 27000, 25000, 17000
])

growth = ((sales - previous_sales) / previous_sales) * 100

print("Growth Percentage:")
print(growth)
```

```python
#11. Monthly Sales
monthly_sales = np.array([
    12000, 15000, 18000, 11000,
    22000, 25000, 17000, 30000,
    28000, 19000, 32000, 35000
])

months = np.array([
    "Jan", "Feb", "Mar", "Apr",
    "May", "Jun", "Jul", "Aug",
    "Sep", "Oct", "Nov", "Dec"
])

for month, sale in zip(months, monthly_sales):
    print(month, sale)
```

```python
#12. Best Month
best_month_index = np.argmax(monthly_sales)

print("Best Month:", months[best_month_index])
print("Highest Sales:", monthly_sales[best_month_index])
```

```python
#13. Worst Month
worst_month_index = np.argmin(monthly_sales)

print("Worst Month:", months[worst_month_index])
print("Lowest Sales:", monthly_sales[worst_month_index])
```

```python
#14. Sort Sales
sorted_sales = np.sort(sales)

print(sorted_sales)
```

```python
#Descending order:
descending_sales = np.sort(sales)[::-1]

print(descending_sales)
```

```python
#15. Unique Values
data = np.array([100, 200, 100, 300, 200, 400, 500, 300])

unique_values = np.unique(data)

print(unique_values)
```

```python
#16. Handle Missing Values
sales_with_nan = np.array([
    12000, 15000, np.nan,
    18000, 22000, np.nan,
    25000
])

print("Mean:", np.nanmean(sales_with_nan))
print("Maximum:", np.nanmax(sales_with_nan))
print("Minimum:", np.nanmin(sales_with_nan))
```

```python
#17. Replace Missing Values
mean_sales = np.nanmean(sales_with_nan)

clean_sales = np.where(
    np.isnan(sales_with_nan),
    mean_sales,
    sales_with_nan
)

print(clean_sales)
```

```python
#39. Outlier Detection
sales = np.array([
    10000, 12000, 15000,
    14000, 16000, 18000,
    20000, 100000
])

Q1 = np.percentile(sales, 25)
Q3 = np.percentile(sales, 75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = sales[
    (sales < lower) |
    (sales > upper)
]

print("Outliers:", outliers)
```
